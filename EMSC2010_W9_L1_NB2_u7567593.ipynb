#### The logistic population model:

$x_{\mathrm{next}} = a x_{\mathrm{initial}}(1-x_{\mathrm{initial}})$

where:
* $x$ is the population size between 0 and 1.
* $x_{\mathrm{initial}}$ is the current population size
* $x_{\mathrm{next}}$ is next year's population size
* $a$ is the ability to reproduce.



```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
### Code the logistic population model
a = 3.6 # a equals fuckability
h = 0.22 # h equals harvesting of a population
x = np.zeros(50)
x[0] = 0.5

for i in range(1,50):
  x[i] = a * x[i-1] * (1 - x[i-1]) - h # The magical calculation
  x[i] = np.maximum(x[i],0)
  #x[i] = np.minimum(x[i],1)

plt.plot(x)
plt.xlabel('Year')
plt.ylabel('Population')
plt.title('Logistic Population Model')
plt.ylim(0,1)
plt.xlim(0,50)
plt.show()
#x_initial = 0.576
#x_next = a * x_initial * (1 - x_initial)
#print(x_next)
```

#### The logistic population model with randomness:

$x_{\mathrm{next}} = (a+r) x_{\mathrm{initial}}(1-x_{\mathrm{initial}})$

where:
* $x$ is the population size between 0 and 1.
* $x_{\mathrm{initial}}$ is the current population size.
* $x_{\mathrm{next}}$ is next year's population size.
* $a$ is the ability to reproduce.
* $r$ is a normally distributed random number that changes each year.

```python
### Code the logistic population model with randomness

a = 3.2 # a equals fuckability
h = 0.1 # h equals harvesting of a population
x = np.zeros(100)
x[0] = 0.9

for i in range(1,100):
  r = np.random.randn() * 0.1 # Random environmental variability
  x[i] = (a + r) * x[i-1] * (1 - x[i-1]) - h # The magical calculation
  x[i] = np.maximum(x[i],0)
  #x[i] = np.minimum(x[i],1)

plt.plot(x)
plt.xlabel('Year')
plt.ylabel('Population')
plt.title('Logistic Population Model')
plt.ylim(0,1)
plt.xlim(0,100)
plt.show()
```
